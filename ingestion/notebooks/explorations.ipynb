# DATA RETRIEVE

```python
import requests
import pandas as pd
from pathlib import Path
from datetime import date

pd.set_option('display.max_colwidth', None)


# URL DATA
URL = "https://data.economie.gouv.fr/api/explore/v2.1/catalog/datasets/prix-des-carburants-en-france-flux-instantane-v2/exports/csv"

# DATA CSV RAW DIR
RAW_DIR =  Path("../../data/raw")
RAW_DIR.mkdir(parents=True, exist_ok=True)

#  CSV DATA PATH
csv_path = RAW_DIR/f"carburants_.csv{date.today()}.csv"

# 
if not csv_path.exists():
    req = requests.get(URL, timeout=210)
    req.raise_for_status()
    csv_path.write_bytes(req.content)

print(csv_path, f"{csv_path.stat().st_size / 1e6:.1f} Mo")

print(csv_path.read_text(encoding="utf-8")[:1000])

# Useleful data
data_to_colllect = ["cp", "adresse", "ville", "gazole_maj", "gazole_prix", "sp95_maj", "sp95_prix", "e85_maj", "e85_prix", "gplc_maj", "gplc_prix", "e10_maj", "e10_prix", "sp98_maj", "sp98_prix"]

df = pd.read_csv(csv_path, sep=";", usecols=data_to_colllect)
df.shape
df
```
